## 1. Đọc dữ liệu và chuẩn bị thư viện

Tải `titanic.csv` vào bảng **Tệp** của Google Colab trước khi chạy.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', 30)

# Tải file titanic.csv của giảng viên lên Colab trước khi chạy.
# Có thể tải tại /content/titanic.csv hoặc /titanic.csv như phiên làm việc trước.
paths = ['/content/titanic.csv', '/titanic.csv', 'titanic.csv', 'titanic3.csv']
titanic_path = next((Path(p) for p in paths if Path(p).is_file()), None)
if titanic_path is None:
    raise FileNotFoundError('Không tìm thấy titanic.csv. Vào thanh Tệp của Colab và tải file lên trước.')

df = pd.read_csv(titanic_path)
df.columns = df.columns.str.strip().str.lower()
df = df.rename(columns={'survival': 'survived'})
for col in ['pclass', 'survived', 'age', 'sibsp', 'parch', 'fare']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
print('Đã đọc:', titanic_path)
print('Kích thước dữ liệu:', df.shape)
assert df.shape == (1309, 14), 'Hãy kiểm tra lại file Titanic do giảng viên cung cấp'
```

## Câu 3 — Tuổi, giới tính và khả năng sống sót

Biểu đồ 1: phân bố số hành khách theo tuổi, tách theo nam/nữ và sống/chết. Biểu đồ 2: **tỷ lệ** sống sót trong từng nhóm tuổi theo giới. Lưu ý **số lượng** khác với **tỷ lệ**: không gọi chiều cao cột đếm là xác suất.

**Viết nhận xét:** so sánh nhóm nữ/nam, trẻ em/người lớn theo *tỷ lệ sống sót*. Kiểm tra số mẫu từng nhóm trước khi kết luận; nhóm quá ít người dễ dao động mạnh.

## 2. Thực hiện bài toán

```python
age_plot = df.dropna(subset=['age','sex','survived']).copy()
g = sns.displot(
    data=age_plot, x='age', hue='survived', col='sex',
    bins=20, multiple='layer', alpha=0.55, height=4, aspect=1.15
)
g.set_axis_labels('Tuổi','Số hành khách')
g.fig.suptitle('Titanic: số hành khách theo tuổi, giới tính, sống sót', y=1.05)
plt.show()

age_plot['age_range'] = pd.cut(age_plot['age'], [0, 10, 18, 30, 40, 60, 100], include_lowest=True)
survival_by_age_sex = age_plot.groupby(['age_range','sex'], observed=True)['survived'].mean().reset_index()
survival_by_age_sex['age_range'] = survival_by_age_sex['age_range'].astype(str)
plt.figure(figsize=(10,5))
sns.barplot(data=survival_by_age_sex, x='age_range', y='survived', hue='sex')
plt.title('Tỷ lệ sống sót theo nhóm tuổi và giới tính')
plt.ylabel('Tỷ lệ sống sót (0 đến 1)')
plt.xlabel('Nhóm tuổi')
plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
display(survival_by_age_sex)
```
