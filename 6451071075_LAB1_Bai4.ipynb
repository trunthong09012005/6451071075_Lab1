## 1. Đọc dữ liệu và chuẩn bị thư viện

Tải `titanic.csv` vào bảng **Tệp** của Google Colab trước khi chạy.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', 30)

# Tải file titanic.csv của giảng viên lên Colab trước khi chạy.
# Có thể tải tại /content/titanic.csv hoặc /titanic.csv như phiên làm việc trước.
paths = ['/content/titanic.csv', '/titanic.csv', 'titanic.csv', 'titanic3.csv']
titanic_path = next((Path(p) for p in paths if Path(p).is_file()), None)
if titanic_path is None:
    raise FileNotFoundError('Không tìm thấy titanic.csv. Vào thanh Tệp của Colab và tải file lên trước.')

df = pd.read_csv(titanic_path)
df.columns = df.columns.str.strip().str.lower()
df = df.rename(columns={'survival': 'survived'})
for col in ['pclass', 'survived', 'age', 'sibsp', 'parch', 'fare']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
print('Đã đọc:', titanic_path)
print('Kích thước dữ liệu:', df.shape)
assert df.shape == (1309, 14), 'Hãy kiểm tra lại file Titanic do giảng viên cung cấp'
```

## Câu 4 — Embarked + Pclass + Sex liên quan Survived

Tính trung bình `survived` trong từng tổ hợp `embarked × pclass × sex`: do `survived` là 0/1, trung bình chính là tỷ lệ sống sót. Tạo 3 biểu đồ theo ba cảng.

**Viết nhận xét:** hạng vé nào có tỷ lệ sống sót cao hơn? Cùng hạng vé, nữ và nam có khác biệt không? Khi so sánh các cảng, đừng quên xem `count`: tỷ lệ 100% của nhóm rất nhỏ không đủ để khái quát.

## 2. Thực hiện bài toán

```python
port_plot = df.dropna(subset=['embarked','pclass','sex','survived']).copy()
port_rate = port_plot.groupby(['embarked','pclass','sex'], observed=True)['survived'].agg(rate='mean',count='size').reset_index()
fig, axes = plt.subplots(1,3,figsize=(15,4),sharey=True)
for ax, port in zip(axes, ['C','Q','S']):
    sub = port_rate[port_rate['embarked'] == port]
    sns.barplot(data=sub, x='pclass', y='rate', hue='sex', ax=ax)
    ax.set_title('Cổng '+port)
    ax.set_xlabel('Loại vé')
    ax.set_ylabel('Tỷ lệ sống sót')
    ax.set_ylim(0,1)
plt.suptitle('Tỷ lệ sống sót theo cổng lên tàu, hạng vé và giới tính',y=1.04)
plt.tight_layout()
plt.show()
print('Bảng tỷ lệ và số lượng (count) để đối chiếu:')
display(port_rate.sort_values(['embarked','pclass','sex']))
```
